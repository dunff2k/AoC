# Advent of Code - Day 11

Find the Day 11 problem [here](https://adventofcode.com/2025/day/11").

## Pre-requisites: specify import(s), set constants, and load data.

```python
PUZZLE_FILENAME = "data/day11.txt"
TEXT_ENCODING = "utf8"

# Node names
OUT = "out"
YOU = "you"
DAC = "dac"
FFT = "fft"
SVR = "svr"
```

```python
def load_data() -> dict[str, list[str]]:
    """Loads the data from the problem into storage structures.
    Example data:
    omi: yzw ulv dyc lvs
    dlj: pjc eow vbm
    kxy: fqs
    phv: jzw
    uhg: jub caw xvt
    aid: ibd gja yxf
    ...
    """
    paths = {}

    with open(PUZZLE_FILENAME, 'r') as file:
        full_file = file.read()
        lines = full_file.strip().split('\n')

    for line in lines:
        start = line[0:3]
        connector_list = line[4:].strip().split()  # Just use split() with no argument!
        paths[start] = connector_list

    return paths
```

# Part One

### Notes
* DFS with memo.

```python
def calculate_number_of_paths() -> int:
    """Calculates the number of paths from YOU to OUT"""

    def helper(start: str) -> int:
        """Helper function to iterate the paths of the graph."""
        if start in memo:
            return memo[start]
        
        connections = data[start]
        if OUT in connections:
            memo[start] = 1 if start not in memo else memo[start] + 1
            return True;
        else:
            sum = 0
            for connection in connections:
                sum += helper(connection)

            return sum
        
    memo = {}
    data = load_data()
    return helper(YOU)
```

```python
print(f"Fewest number of button presses: {calculate_number_of_paths()}")
```

# Part Two

```python
def calculate_number_of_paths_via_dac_and_fft() -> int:
    """Calculates the number of paths from YOU to OUT via DAC and FFT"""

    def helper(start: str, previously_visited_dac: bool, previously_visited_fft: bool) -> int:
        """Helper function to iterate the paths of the graph."""
        memo_key = (start, previously_visited_dac, previously_visited_fft)
        
        if memo_key in memo:
            return memo[memo_key]
        
        dac = previously_visited_dac or (start == DAC)
        fft = previously_visited_fft or (start == FFT)

        if start == OUT:
            result = 1 if (dac and fft) else 0
            return result;
        
        # Recursive step.
        sum = 0
        connections = data[start]
        for connection in connections:
            sum += helper(connection, dac, fft)
        
        memo[memo_key] = sum
        return sum
        
    memo = {}
    data = load_data()
    return helper(SVR, False, False)
```

```python
print(f"Number of paths to OUT containing DAC and FFT: {calculate_number_of_paths_via_dac_and_fft()}")
```
