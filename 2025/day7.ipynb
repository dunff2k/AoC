## Pre-requisites: specify import(s), set constants, and load data.

````python
PUZZLE_FILENAME = "data/day7.txt"
TEXT_ENCODING = "utf8"
SPLITTER_CHAR = "^"
START_CHAR = "S"
````

````python
def load_data() -> list[str]:
    """Loads the data into a grid
    
    Example of structure:
    ['......................................................................S......................................................................',
     '......................................................................^......................................................................',
     ...
    ]
    
    Returns:
        The grid of characters.
    """
    grid = []
    with open(PUZZLE_FILENAME, mode="r", encoding=TEXT_ENCODING) as file:
        grid = file.read().splitlines()
    return grid
````

# Part One

### Notes

You quickly locate a diagram of the tachyon manifold (your puzzle input). A tachyon beam enters the manifold at the location marked S; tachyon beams always move downward. Tachyon beams pass freely through empty space (.). However, if a tachyon beam encounters a splitter (^), the beam is stopped; instead, a new tachyon beam continues from the immediate left and from the immediate right of the splitter.

For example:
```
.......S.......
...............
.......^.......
...............
......^.^......
...............
.....^.^.^.....
...............
....^.^...^....
...............
...^.^...^.^...
...............
..^...^.....^..
...............
.^.^.^.^.^...^.
...............
```

````python
def count_beam_splits() -> int:
    """Count the number of beam splits through the grid.
    
    Returns:
        The number of times the beam splits.
    """
    cache = set()
    splitter_count = 0
    grid = load_data()
    start_ray = grid[0].index(START_CHAR)
    assert len(grid) > 0 and len(grid[0]) > 0

    def process_level(level: int, index: int):
        nonlocal splitter_count
        assert index >= 0 and index < len(grid[0])

        if level == len(grid) - 1:
            return 1
        if (level, index) in cache:
            return 0
        
        the_char = grid[level][index]
        cache.add((level, index))

        if the_char == SPLITTER_CHAR:
            splitter_count += 1
            process_level(level + 1, index + 1)
            process_level(level + 1, index - 1)   
        else:
            process_level(level + 1, index)
        
    process_level(1, start_ray)
    return splitter_count    
````

## Solution to Part One 

````python
print(f"Number of beam splits: {count_beam_splits()}")
````
