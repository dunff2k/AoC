# Advent of Code - Day 5

Find the Day 5 problem [here](https://adventofcode.com/2025/day/5").

## Pre-requisites: specify import(s), set constants, and load data.

```python
# Replace filename with actual filename.
PUZZLE_FILENAME = "data/day5.txt"
```

```python
def parse_range(line: str) -> tuple[int, int]:
    """Helper function to convert ranges to int tuples."""
    RANGE_DELIMITER = "-"
    begin, end = line.split(RANGE_DELIMITER)
    return int(begin), int(end)
```

```python
def load_data() -> tuple[list[tuple[int, int]], list[int]]:
    """Fetches the puzzle input data.

    File is in a tricky format i.e. a mix of `ranges` 
    and Ids separated by newline:

    51716793013563-58576571790780
    416293147622133-418544898567771
    204293845440550-209480228080573
    475865041061127-476366294808733
    142458197886533-142780858100993

    329928111646776
    482309026155231
    525241827868042
    96135717389080
    499110589511466

    Returns:
        A list of ranges of product IDs that are fresh, and list of 
        product IDs.
    """
    TEXT_ENCODING = "utf-8"
    with open(PUZZLE_FILENAME, "r", encoding=TEXT_ENCODING) as file:
        the_text_file = file.read()
        ranges_str, ids_str = the_text_file.split("\n\n")
        fresh_id_ranges = [parse_range(line) for line in ranges_str.splitlines()]
        ids = [int(line) for line in ids_str.splitlines()]    
    return fresh_id_ranges, ids
```

## Part One

### Notes

* The fresh ID ranges are inclusive: the range 3-5 means that ingredient IDs 3, 4, and 5 are all fresh.
* The ranges can also overlap; an ingredient ID is fresh if it is in any range.
* Which of the available ingredient IDs are fresh?

```python
def count_fresh_ids() -> int:
    fresh_id_ranges, product_ids = load_data()

    return sum(
        any(begin <= id <= end for begin, end in fresh_id_ranges)
        for id in product_ids
    )
```

## Solution for Part One

```python
print(f"Number of fresh IDs: {count_fresh_ids()}")
```

# Part Two

### Notes
* Need to know all of the IDs that the fresh ingredient ID ranges consider to be fresh. 
* An ingredient ID is still considered fresh if it is in any range.

```python
def count_all_fresh_product_ids():
    """Count all the fresh product IDs in a given range."""
    fresh_product_ranges, _ = load_data()
    sorted_ranges = sorted(fresh_product_ranges, key=lambda x: x[0])
    
    # Maintain a list of all valid intervals
    merged = []
    for start, end in sorted_ranges:
        # If the new interval overlaps or is adjacent.
        if merged and start <= merged[-1][1] + 1:
            merged[-1] = (merged[-1][0], max(merged[-1][1], end))
        else:
            merged.append((start, end))
    return sum(end - start + 1 for start, end in merged)
               

```

## Solution for Part Two

```python
print(f"{count_all_fresh_product_ids()}")
```
