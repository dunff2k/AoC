# Advent of Code - Day 6

Find the Day 6 problem [here](https://adventofcode.com/2025/day/6").

## Pre-requisites: specify import(s), set constants, and load data.

```python
import numpy as np
from math import prod

# Replace filename with actual filename.
PUZZLE_FILENAME = "data/day6.txt"
TEXT_ENCODING = "utf8"
PLUS_OPERATOR = "+"
```

```python
def load_data() -> tuple[list[str], list[str]]:
    """Load data for arithmetic.
    
    Tricky data format since arithmetic takes place vertically:
    123 328  51 64 
    45 64  387 23 
    6 98  215 314
    *   +   *   +  
    """
    with open(PUZZLE_FILENAME, "r", encoding=TEXT_ENCODING) as file:
        data = [line.split() for line in file.read().splitlines()]
    operations = data.pop()
    return data, operations
```

# Part One

### Notes
* Each number is given in its own column, with the most significant digit at the top and the least significant digit at the bottom. (Problems are still separated with a column consisting only of spaces, and the symbol at the bottom of the problem is still the operator to use.)

```python
def calculate_sum_of_column_arithmetic() -> int:
    """Calculate the sum of the column arithmetic."""
    numbers, operations = load_data()
    numbers = np.array(numbers, dtype=int)
    assert len(numbers[0]) == len(operations)
    sum = 0
    for i, op in enumerate(operations):
        match op:
            case "*":
                sum += np.prod(numbers[:, i])
            case "+":
                sum += np.sum(numbers[:, i])
    return sum
```

## Solution for Part One

```python
print(f"Total is: {calculate_sum_of_column_arithmetic()}")
```

# Part Two

### Notes
* Each number is given in its own column, with the most significant digit at the top and the least significant digit at the bottom. (Problems are still separated with a column consisting only of spaces, and the symbol at the bottom of the problem is still the operator to use.)

```python
def load_data_and_get_columns() -> tuple[list[str], list[str]]:
    """Load data for arithmetic, pad text, and split into columns.
    
    Tricky data format since arithmetic takes place vertically:
    123 328  51 64 
    45 64  387 23 
    6 98  215 314
    *   +   *   +  
    """
    with open(PUZZLE_FILENAME, "r", encoding=TEXT_ENCODING) as file:
        raw_data = file.read().splitlines()
    
    operations = raw_data.pop().split()

    max_len = max(len(line) for line in raw_data)
    the_data = [line.ljust(max_len) for line in raw_data]

    columns = []
    for col_i in range(max_len - 1, -1, -1):
        col = ''.join(line[col_i] for line in the_data)
        columns.append(col)

    return columns, operations
```

```python
def get_problems(columns: list[str]) -> list[list[str]]:
    """Split the columns into specific problems to solve."""
    problems = []
    current_problem = []
    
    for col in columns:
        if col.strip() == '':
            if current_problem:
                problems.append(current_problem)
                current_problem = []
        else:
            current_problem.append(col)
    if current_problem:
        problems.append(current_problem)
    return problems
```

```python
def get_sum_of_advanced_problems() -> int:
    """Calculate the sum of the problems."""
    columns, operations = load_data_and_get_columns()
    problems = get_problems(columns)
    
    total = 0
    
    for prob_cols in problems:
        operator = operations.pop()
        
        # Build numbers from each column.
        numbers = []
        for col in prob_cols:
            num_str = ''.join(col_text for col_text in col if col_text.isdigit())
            if num_str:
                numbers.append(int(num_str))
        
        result = sum(numbers) if operator == PLUS_OPERATOR else prod(numbers)
        total += result
    
    return total
```

## Solution to Part Two

```python
print(f"{get_sum_of_advanced_problems()}")
```
