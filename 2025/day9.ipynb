Find the Day 9 problem [here](https://adventofcode.com/2025/day/9").

## Pre-requisites: specify import(s), set constants, and load data.

```python
PUZZLE_FILENAME = "data/day9.csv"
TEXT_ENCODING = "utf8"
```

```python
def load_data() -> list[tuple[int, int]]:
    """Loads the data into a list

    Example data:
    65306,95530
    64042,95530
    64042,96451
    63047,96451
    63047,96684
    61845,96684
    61845,96979
    60661,96979
    """

    with open(PUZZLE_FILENAME, mode="r", encoding=TEXT_ENCODING) as file:
        vals = [tuple(map(int, line.rstrip().split(","))) for line in file]
        swapped = [ (y, x) for (x, y) in vals ]
    return vals
```

# Part One

```python
def calculate_largest_rectangle_area() -> int:
    """Calculates the largest possible rectangle.
    
    Returns:
        The area of the largest possible rectangle.
    """
    red_tiles = load_data()
    max_area = float("-inf")

    for current_row, current_col in red_tiles:
        for row, col in red_tiles:

            if row == current_row and current_col == col:
                continue

            area = abs(current_row - row + 1) * abs(current_col - col + 1)
            max_area = max(max_area, area)

    return max_area
```

### Get Solution to Part One

```python
print(f"Solution to part one: {calculate_largest_rectangle_area()}")
```

# Part Two

```python
from shapely.geometry import Polygon, box

def largest_rect_area_red_and_green_tiles() -> int:
    """Calculates the largest possible rectangle composed of only red and green tiles.

    Leverages shapely library for polygon functions such as raycasting.

    Returns:
        Area of largest rectangle that can be composed of red and green tiles.
    """
    tiles = load_data()
    polygon = Polygon(tiles)
    
    max_area = 0
    
    # Try all pairs of red tiles as opposite corners.
    for i, (x1, y1) in enumerate(tiles):
        for (x2, y2) in tiles[i+1:]:
            x_min, x_max = min(x1, x2), max(x1, x2)
            y_min, y_max = min(y1, y2), max(y1, y2)
            
            area = (x_max - x_min + 1) * (y_max - y_min + 1)
            if area <= max_area:
                continue
            
            # Shrink the rectangle slightly to see if the rect is in the polygon.
            # Avoids boundary issues.
            rect = box(x_min + 0.1, y_min + 0.1, x_max - 0.1, y_max - 0.1)
            
            if polygon.contains(rect):
                max_area = area
    
    return max_area
```

### Get Solution to Part Two

```python
print(f"The solution to part two is: {largest_rect_area_red_and_green_tiles()}")
```
