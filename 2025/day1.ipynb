# Advent of Code - Day 1

Find the Day 1 problem [here](https://adventofcode.com/2025/day/1").

## Pre-requisites: specify import(s), set constants, and load data.

```python
NUMBERS_ON_DIAL = 100
STARTING_POSITION = 50
TURN_RIGHT = "R"
TURN_LEFT = "L"

# Change this to reflect your filename.
PUZZLE_FILENAME = "data/day1.txt" 
```

```python
def load_data() -> tuple[list[str], list[str]]:
    directions = []
    nums = []

    with open(PUZZLE_FILENAME, 'r') as file:
        lines = [line.strip() for line in file]
        
    directions = [line[0] for line in lines]
    nums = [line[1:] for line in lines]

    return directions, nums
```

# Part One

```python
def get_password() -> int:
    """Get the number of zeros resulting from the input data.
    
    Returns:
        The number of zeros landed on while following the 
        supplied instructions.
    """
    zero_count = 0
    current_point = STARTING_POSITION
    direction, number = load_data()
    for (dir, num) in zip(direction, number):
        distance = int(num) if dir == TURN_RIGHT else -int(num)
        current_point = (current_point + distance) % NUMBERS_ON_DIAL

        if (current_point == 0):
            zero_count += 1

    return zero_count
```

### Get Solution to Part One

```python
print(f"The number of zeros counted: {get_password()}")
```

# Part Two

```python
def get_password_new_policy() -> int:
    """The correct password according to the second policy.
    
    The new policy requires counting the total number of zeroes 
    crossed AT ANY TIME in the duration of the turns. 
    Not only those turns that end in zero.
    
    Returns:
        The number of zeros crossed while entering the numbers.
    """
    zero_count = 0
    current_point = STARTING_POSITION
    direction, number = load_data()
    for turn_dir, num in zip(direction, number):
        distance = int(num)
        if turn_dir == TURN_RIGHT:
            zero_count += (current_point + distance) // NUMBERS_ON_DIAL
            current_point = (current_point + distance) % NUMBERS_ON_DIAL
        else: 
            if current_point == 0:
                zero_count += distance // NUMBERS_ON_DIAL
            elif distance >= current_point:
                zero_count += (distance - current_point) // NUMBERS_ON_DIAL + 1
            
            current_point = (current_point - distance) % NUMBERS_ON_DIAL

    return zero_count
```

### Get Solution to Part Two

```python
print(f"The number of zeroes counted: {get_password_new_policy()}")
```
