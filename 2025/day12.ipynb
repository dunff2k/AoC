## Pre-requisites: specify import(s), set constants, and load data.

```python
import numpy as np
PUZZLE_FILENAME = "data/day12.txt"
TEXT_ENCODING = "utf8"
```

```python
def load_data() -> tuple[list[np.ndarray], list[tuple[int, int]], tuple]:
    """Loads the data from the problem into storage structures.
    
    Example data:
    3:
    ###
    .##
    ..#

    4:
    ##.
    .##
    ..#

    48x44: 56 56 65 51 44 50
    49x36: 35 32 21 36 37 31
    39x50: 29 40 31 31 40 37
    48x49: 43 44 44 32 53 40
    ...
    """
    with open(PUZZLE_FILENAME, 'r') as file:
        full_file = file.read().splitlines()
    
    presents_temp = [line for line in full_file if "x" not in line]
    regions_temp = [line for line in full_file if "x" in line]
    
    presents = segment_presents(presents_temp)
    dimensions = segment_dimensions(regions_temp)
    quantities = segment_quantities(regions_temp)

    return presents, dimensions, quantities


def segment_dimensions(raw_regions_data: list[str]) -> list[tuple[int, int]]:
    """Extract the dimensions of the grids from the lines of text."""
    dims = []
    for item in raw_regions_data:
        parts = item.split(":")
        width, height = parts[0].split("x")
        dims.append((int(height), int(width)))  
    return dims


def segment_quantities(raw_regions_data: list[str]) -> list[tuple]:
    """Extract the quantities of each present required from a line of text"""
    quant = []
    for item in raw_regions_data:
        parts = item.split(":")
        quants_temp = parts[1].strip().split(" ")
        quant.append(tuple(int(x) for x in quants_temp)) 
    return quant

def segment_presents(raw_presents_data: list[str]) -> list[np.ndarray]:
    """Extract the present shapes from the line of text."""
    presents = []
    temp = []

    for item in raw_presents_data:
        if ":" in item:
            temp = []
        elif not item:
            if temp:  
                presents.append(np.array([[1 if c == '#' else 0 for c in row] for row in temp]))
                temp = []
        else:
            temp.append(item)

    return presents
```

# Part One

### Notes
* The problem appears to have a 'trolling' aspect to it since the complicated transformations suggested in the example were not necessary to solve the actual problem. The test data was very kind.

```python
def calculate_number_of_fitting_regions() -> int:
    """Calculate the number of regions that could successfully hold the required gifts."""
    presents, dimensions, quantities = load_data()
    
    shape_areas = [np.sum(shape) for shape in presents]
    
    fitting_count = 0
    
    for (rows, cols), quant in zip(dimensions, quantities):
        region_area = int(rows) * int(cols)
        
        # It appears that checking whether the total space available 
        # is sufficient is actually sufficient to solve this problem.
        needed_area = sum(quant[i] * shape_areas[i] for i in range(len(quant)))
        
        if needed_area <= region_area:
            fitting_count += 1
    
    return fitting_count
```

### Get Solution to Part One

```python
print(f"Number of fitting regions: {calculate_number_of_fitting_regions()}")
```
