## Pre-requisites: specify import(s), set constants, and load data.

```python
import csv

# Change this to your own filename.
PUZZLE_FILENAME = "data/day2.csv" 
```

```python
def load_data() -> list[str]:
    """Load data from puzzle file."""
    rows = []

    with open(PUZZLE_FILENAME, 'r') as file:
        reader = csv.reader(file)
        rows = next(reader) 

    return rows
```

## Part One

```python
def get_ranges(data: list[str]) -> list[tuple[int, int]]:
    """Splits the data into (start, end) tuples."""
    return [(int(start), int(end)) for row in data for start, end in [row.split('-')]]
```

```python
def is_mirror_number(product_id: int) -> bool:
    """Checks whether supplied product ID is a repeated number"""
    str_n = str(product_id)
    half = len(str_n) // 2
    return len(str_n) % 2 == 0 and str_n[:half] == str_n[half:]
```

```python
def calculate_sum_invalid_product_ids() -> int:
    """Calculates whether product ID is invalid.
    
    Identifies invalid product IDs which are IDs made only 
    of some sequence of digits repeated twice. So, 55 (5 twice), 
    6464 (64 twice), and 123123 (123 twice) would all be invalid IDs.
    
    Returns: 
        The sum of the invalid product IDs.
    """
    data = load_data()
    ranges = get_ranges(data)
    sum = 0
    for start, end in ranges:
        for i in range(int(start), int(end) + 1):
                if is_mirror_number(i):
                    sum += i
    return sum
```

## Solution to Part One.

```python
print(f"Number of invalid IDs: {calculate_sum_invalid_product_ids()}")
```

# Part Two

```python
def sum_invalid_product_ids_new_policy() -> int:
    """Sums the invalid product IDs according to repitition-based policy. 
    
    An ID is invalid if it is made only of some sequence of digits repeated 
    at least twice. So, 12341234 (1234 two times), 123123123 (123 three times), 
    1212121212 (12 five times), and 1111111 (1 seven times) are all invalid IDs.
    
    Returns:
        The sum of the invalid product IDs.
    """
    data = load_data()
    ranges = get_ranges(data)

    max_val = max(end for _, end in ranges)
    max_digits = len(str(max_val))

    invalid_ids = set()

    for pattern_len in range(1, max_digits // 2 + 1):  
        start = 1 if pattern_len == 1 else 10 ** (pattern_len - 1)
        
        for pattern in range(start, 10 ** pattern_len):
            s = str(pattern)
            repeated = s + s  
            while len(repeated) <= max_digits:
                num = int(repeated)
                if num <= max_val:
                    invalid_ids.add(num)
                repeated += s

    # Sum the product IDs that fall into any range.
    total = sum(
        inv for inv in invalid_ids
        if any(start <= inv <= end for start, end in ranges)
    )

    return total
```

## Solution to Part Two

```python
print(f"New sum of product IDs: {sum_invalid_product_ids_new_policy()}")
```
