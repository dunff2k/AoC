# Advent of Code - Day 3

Find the Day 3 problem [here](https://adventofcode.com/2025/day/3").

## Pre-requisites: specify import(s), set constants, and load data.

```python
# Change this to your own filename.
PUZZLE_FILENAME = "data/day3.txt" 
```

```python
def load_data() -> list[str]:
    """Load the data for the task."""
    batteries = []
    with open(PUZZLE_FILENAME, 'r') as file:
        for line in file:
            batteries.append(line.strip())

    return batteries
```

# Part One

```python
def pick_two_batteries() -> int:
    """Select two batteries to give the maximum power."""
    batteries = load_data()
    total = 0
    for battery in batteries:
        first_part = battery[:-1]
        max_first = max(first_part)
        max_first_idx = first_part.index(max_first)
        max_second = max(battery[max_first_idx + 1:])    
        total += int(max_first + max_second)

    return total
```

## Solution to Part One

```python
print(f"The total output joltage: {pick_two_batteries()}")
```

# Part Two

```python
def select_batteries_from_bank(bank: list[str]):
    """Selects the optimal batteries from the battery bank."""
    NUMBER_OF_BATTERIES_TO_PICK = 12
    BANK_LENGTH = len(bank)
    digits = ""
    last_pick = -1
    for i in range(0, NUMBER_OF_BATTERIES_TO_PICK):
        start = last_pick + 1
        last_possible = (BANK_LENGTH - NUMBER_OF_BATTERIES_TO_PICK) + i + 1
        to_search = bank[start: last_possible]
        largest = max(to_search)
        last_pick = start + to_search.index(largest)
        digits += largest
    return digits

    
```

```python
def pick_twelve_batteries() -> int:
    """Get the maximum joltage for picking 12 batteries
    
    Searches the bank until the last feasible location for each
    battery given its position in the sequence. 
    
    Returns:
        The total joltage.
    """
    batteries = load_data()
    total = 0
    for bank in batteries:
        total += int(select_batteries_from_bank(bank))

    return total
```

## Solution to Part Two

```python
print(f"Total output joltage: {pick_twelve_batteries()}")
```
